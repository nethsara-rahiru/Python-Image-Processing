```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import cv2
import PIL
from PIL import Image
import scipy
import skimage
import ipywidgets

print("All packages imported successfully!\n")
print(f"NumPy        : {np.__version__}")
print(f"Matplotlib   : {matplotlib.__version__}")
print(f"OpenCV       : {cv2.__version__}")
print(f"Pillow (PIL) : {PIL.__version__}")
print(f"SciPy        : {scipy.__version__}")
print(f"scikit-image : {skimage.__version__}")
print(f"ipywidgets   : {ipywidgets.__version__}")

# Quick functional test - load a built-in sample image and display it
from skimage import data
test_img = data.camera()
plt.imshow(test_img, cmap='gray')
plt.title("Setup check: if you can see this image, you're ready!")
plt.axis('off')
plt.show()
```
