```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y =  np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show
```

```python
# Creating a Random Array

random_array = np.random.randint(0, 256, (100, 100))
print(random_array.shape)

# Display the array as an image using a gray color map
plt.imshow(random_array, cmap='gray')
plt.title("Random Grayscale Array")
plt.colorbar()
plt.show()
```

```python
arr_a = np.full((50,50), 50)
arr_b = np.full((50,50), 100)
arr_c = np.full((50,50), 200)


plt.figure(figsize=(8, 4))

plt.subplot(1,3,1)
plt.imshow(arr_a, cmap = 'gray', vmin = 0, vmax = 255)
plt.title(50)

plt.subplot(1,3,2)
plt.imshow(arr_b, cmap = 'gray', vmin = 0, vmax = 255)
plt.title(10)

plt.subplot(1,3,3)
plt.imshow(arr_c, cmap = 'gray', vmin = 0, vmax = 255)
plt.title(200)
```
