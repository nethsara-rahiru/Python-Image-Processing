```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

```python
img_pil=Image.open("image.jpg")
img_pil_array=np.array(img_pil)
```

```python
print(img_pil_array.shape)
```

```python
print(img_pil_array.dtype)
```

```python
plt.imshow(img_pil_array)
plt.axis("off")
plt.show
```

```python
img_cv=cv2.imread("image.jpg")
```

```python
print(img_cv.shape)
print(img_cv.dtype)
```

```python
plt.imshow(img_cv)
plt.show
```

```python
img_cv_rgb=cv2.cvtColor(img_cv,cv2.COLOR_BGR2RGB)
plt.imshow(img_cv_rgb)
plt.show
```

```python
img_gray=cv2.imread("image.jpg",cv2.IMREAD_GRAYSCALE)
print(img_gray.shape)
```

```python
plt.imshow(img_gray,cmap="gray")
plt.axis("off")
plt.show()
```

```python
forced_gray=cv2.imread("image.jpg",cv2.IMREAD_GRAYSCALE)
plt.imshow(forced_gray,cmap="gray")
plt.axis("off")
plt.show
```

```python
cv2.imwrite("Saved_img.jpg", forced_gray)
```
