```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
image_path = "image.jpg"
img = np.array(Image.open(image_path))
```

```python
plt.figure(figsize=(6,5))
if img.ndim == 2:
    plt.imshow(img,cmap="gray")
else:
    plt.imshow(img)
plt.title("Image")
plt.axis("off")
plt.show()
```

```python
print(img.shape)
print(img.dtype)
print(img.max())
print(img.min())
```

```python
height = img.shape[0]
width = img.shape[1]
dimentions = img.shape[2]
print(height, width, dimentions)
```

```python
if img.ndim == 3:
    print("Color Image")
else:
    print("Grayscale Image")
```

```python
red_channel = img[:,:,0]
green_channel = img[:,:,1]
blue_channel = img[:,:,2]
```

```python
plt.figure(figsize=(5,4))

plt.subplot(1,3,1)
plt.imshow(red_channel,cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(green_channel,cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(blue_channel,cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()
```
