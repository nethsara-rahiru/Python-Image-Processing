```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("image1.png", cv2.IMREAD_GRAYSCALE).astype(np.float32)
```

```python
plt.figure(figsize=(6,5))
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Orginal Image")
plt.axis("off")
```

```python
brighter = img + 60
brighter = np.clip(brighter, 0, 255)
brighter = brighter.astype(np.uint8)
```

```python
darker = img - 60
darker = np.clip(darker, 0, 255)
darker = darker.astype(np.uint8)
```

```python
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.imshow(darker, cmap="gray", vmin=0, vmax=255)
plt.title("-(60)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Referance")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(brighter, cmap="gray", vmin=0, vmax=255)
plt.title("+(60)")
plt.axis("off")
```

```python
plt.figure(figsize=(15, 4))

plt.subplot(1, 4, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("-(60)")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Referance")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("+(60)")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("+(60)")
plt.axis("off")
```

```python
minVale = img.min()
maxValue = img.max()
```

```python
stretched = (img - minVale) * (255.0 / (maxValue - minVale))
stretched = stretched.astype(np.uint8)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretched, cmap='gray')
plt.title("Contrast Stretched")
plt.axis("off")
```

```python
nagitive = 255 - img.astype(np.uint8)
```

```python
nagitive_cv2 = cv2.bitwise_not(img.astype(np.uint8))
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(nagitive, cmap='gray')
plt.title("Contrast Stretched")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(nagitive_cv2, cmap='gray')
plt.title("Cv2 Stretched")
plt.axis("off")
```

```python
_, bin_img_80 = cv2.threshold(img, 100, 105, cv2.THRESH_BINARY)
```

```python
_, bin_img_128 = cv2.threshold(img, 128, 255, cv2.THRESH_BINARY)
```

```python
_, bin_img_180 = cv2.threshold(img, 100, 200, cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(bin_img_80, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(bin_img_128, cmap='gray')
plt.title("Contrast Stretched")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(bin_img_180, cmap='gray')
plt.title("Cv2 Stretched")
plt.axis("off")
```
