```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("cameraman.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(img, cmap="gray")
plt.title("Orginal Image")
plt.axis("off")
plt.show()
```

```python
laplacian_kernel = np.array([
    [0, 1, 0],
    [1,-4, 1],
    [0, 1, 0]], dtype=np.float32)
```

```python
laplacian_manual = cv2.filter2D(img.astype(np.float32), -1, laplacian_kernel)
laplacian_manual_display = cv2.convertScaleAbs(laplacian_manual)
```

```python
laplacian_cv2 = cv2.Laplacian(img, cv2.CV_64F, ksize=3)
laplacian_cv2_display = cv2.convertScaleAbs(laplacian_cv2)
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap="gray")
plt.title("Orginal Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(laplacian_manual_display, cmap="gray")
plt.title("Manual Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(laplacian_cv2_display, cmap="gray")
plt.title("cv2 Image")
plt.axis("off")

plt.show()
```

```python
sobel_kernel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1]],dtype = np.float32)
```

```python
sobel_kernel_y = np.array([
    [-1, -2,-1],
    [-0, 0, 0],
    [ 1, 2, 1]],dtype = np.float32)
```

```python
sobel_x_manual = cv2.filter2D(img.astype(np.float32), -1, sobel_kernel_x)
sobel_y_manual = cv2.filter2D(img.astype(np.float32), -1, sobel_kernel_y)

sobel_manual_magnitude = cv2.magnitude(sobel_x_manual, sobel_y_manual)
sobel_manual_display = cv2.convertScaleAbs(sobel_manual_magnitude)
```

```python
sobel_x_cv2 = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y_cv2 = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

sobel_cv2_magnitude = cv2.magnitude(sobel_x_cv2, sobel_y_cv2)
sobel_cv2_display = cv2.convertScaleAbs(sobel_cv2_magnitude)
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap="gray")
plt.title("Orginal Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(sobel_manual_display, cmap="gray")
plt.title("Sobel Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sobel_cv2_display, cmap="gray")
plt.title("cv2 Image")
plt.axis("off")

plt.show()
```

```python
sharpening_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]], dtype = np.float32)
```

```python
sharpened_manual =cv2.filter2D(img, -1, sharpening_kernel)
```

```python
plt.figure(figsize=(15,4))

plt.subplot(1,2 ,1)
plt.imshow(img, cmap="gray")
plt.title("Orginal Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(sharpened_manual, cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.show()
```
